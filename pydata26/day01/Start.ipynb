# 리스트와 반복구조 연습

```python
num_list = [23, 45, 27, 11, 25, 65, 78]
#getindaex(num_list, target)
def getindaex(num_list, target):
    a=0
    for i in range(len(num_list)):
        if target == num_list[i]:
            return i
print(getindaex(num_list, 25))

#getMax(num_list)
def getMax(num_list):
    max = num_list[0]
    for i in range(len(num_list)):
        if max < num_list[i]:
            max = num_list[i]
    return max
print(getMax(num_list))

#getMin(num_list)
def getMin(num_list):
    min = num_list[0]
    for i in range(len(num_list)):
        if min > num_list[i]:
            min = num_list[i]
    return min
print(getMin(num_list))

#countGT(num_list, target)
def countGT(num_list, target):
    a=0
    for i in range(len(num_list)):
        if target < num_list[i]:
            a += 1
    return a
print(countGT(num_list, 42))

#sumList(num_list)
def sumList(num_list):
    a=0
    for i in range(len(num_list)):
        a += num_list[i]
    return a
print(sumList(num_list))

#swapList(num_list)
def swapList(num_list):
    for i in range(len(num_list) // 2):
        index = len(num_list) - 1 - i
        num_list[i], num_list[index] = num_list[index], num_list[i]
    return num_list
print(swapList(num_list))
```
